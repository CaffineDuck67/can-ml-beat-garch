# 01 - Exploratory Data Analysis
Loads the cleaned daily panel and the target/feature tables and shows the raw material the benchmark forecasts.

```
import numpy as np, pandas as pd, matplotlib.pyplot as plt
import sys; sys.path.insert(0, '..')
from src import utils
utils.configure_matplotlib()
cfg = utils.load_config('../config.yaml')
clean = utils.read_table(utils.CLEAN_DAILY)
feats = utils.read_table(utils.FEATURES)
targets = utils.read_table(utils.TARGETS)
print(clean.shape, feats.shape, targets.shape)
```

```
clean[[utils.SP500_CLOSE, utils.VIX]].plot(subplots=True, figsize=(10,6), title='S&P 500 and VIX')
plt.tight_layout()
```

```
for h in cfg['targets']['horizons']:
    col = utils.rv_target_col(h)
    targets[col].plot(figsize=(10,4), title=f'Realized vol target {col} (%)')
plt.tight_layout()
```

```
plt.figure(figsize=(10,4))
plt.plot(clean.index, clean[utils.SP500_RET]*100, lw=0.6)
plt.title('Daily S&P 500 log returns (%)'); plt.tight_layout()
```

```
print('Feature summary:')
feats.describe().T[['mean','std','min','max']]
```

```
# Correlation of candidate features with the h=21 realized-vol target
t21 = targets[utils.rv_target_col(21)]
corr = feats.apply(lambda c: c.corr(t21.reindex(feats.index)))
corr.sort_values(key=abs, ascending=False).plot(kind='barh', figsize=(8,6))
plt.title('|corr| with RV_21'); plt.tight_layout()
```
